```python
import pandas as pd
import os
```

```python
KEYWORD = "simulation"
simulation_files = [x for x in os.listdir() if x.startswith(KEYWORD)]
```

```python
master_df = pd.read_csv(simulation_files[0])

files_to_merge = simulation_files[1:]
```

```python
print(f"Files to merge: {len(files_to_merge)}")

for file in files_to_merge:

    print(f"Loading file '{file}'")
    df_new = pd.read_csv(file)

    print("Merging...")
    master_df = pd.concat([master_df, df_new])
```

```python
MASTER_FILE = "master_results.csv"

master_df.to_csv(MASTER_FILE)
print(f"File saved to '{MASTER_FILE}'")
```
